```python
### GUESS THE NUMBER GAME ###

import random as rand

lost = True
pc_pick = rand.randint(1, 21)

for i in range(0, 5):
    usr_pick = int(input("Guess what number I picked? "))
    if pc_pick == usr_pick:
        lost = False
        print("You got it!".center(40, "-"))
        break
    elif pc_pick < usr_pick:
        print("Guess lower".center(40, "-"))
        continue
    else:
        print("Guess higher".center(40, "-"))
        continue
if lost:
    print("Oh no! No more guesses left!")
```

```python
###PERSONAL LIBRARY MANAGEMENT SYSTEM###


def add(li):
    # افزودن عنوان و نام نویسنده
    print("You are in 'add' interface.")
    book_title = input("Enter book title. ")
    if book_title.lower().capitalize() in lib:
        print("Book already exists.")
        return
    else:
        auth_name = input("Enter name of the author. ")
        li["Title"] = book_title.lower().capitalize()
        li["Author"] = auth_name.lower().capitalize()
        print(li)
        return li


def show(li):
    # همه اطلاعات موجود در دیکشنری    
    if not len(li):
        print("No Books in this library. Try adding more.")
        return
    if len(li):
        names, auths = li["Title"], li["Author"]
        print(names, auths)


def search(li):
    #گشتن در دیکشنری 
    book_title = input("Enter book title:").lower().capitalize()
    for book in lib:
        if li[book] == book_title:
            print(f"Author name: {li["Author"]}")
            return
        else:
            continue
    print("Not found.")
    

lib = {}
#اگر دیکشنری گلوبال هم بود چیز خوبی میشد
print("WELCOME TO THIS LIBRARY".center(60, '*'))
while True:
    usr_inpt = input("Please pick an option:\n1.Add a new book\n2.Show book info\n3.Search for a book\n4.exit")

#انتخابای کاربر رو مپ میکنیم
    if usr_inpt == '4':
        break
    if usr_inpt == '1':
       lib =  add(lib)
    if usr_inpt == '2':
        show(lib)
    if usr_inpt == '3':
        search(lib)
```
