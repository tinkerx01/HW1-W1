```python
### GUESS THE NUMBER GAME ###

import random as rand

lost = True
pc_pick = rand.randint(1, 21)

for i in range(5):
    usr_pick = int(input("Guess what number I picked? "))
    if pc_pick == usr_pick:
        lost = False
        print("You got it!".center(40, "-"))
        break
    elif pc_pick < usr_pick:
        print("Guess lower".center(40, "-"))
        continue
    else:
        print("Guess higher".center(40, "-"))
        continue
if lost:
    print("Oh no! No more guesses left!")
```

```python
###PERSONAL LIBRARY MANAGEMENT SYSTEM###


def add(li):
    # افزودن عنوان و نام نویسنده
    print("You are in 'add' interface.")
    for i in range(3):
        #سه بار می تونید تلاش کنید
        title = input("Enter book title: ").lower().capitalize()
        if not title.replace(" ", "").isalnum():
            print("English letters only.")
        elif title in li:
            print("Book already exists.")
            return 
        else: 
            break

    for i in range(3):
        auth_name = input("Enter name of the author: ").lower().capitalize()
        if not auth_name.replace(" ", "").isalpha():
            print("English letters only.")
        else:
            li[title] = auth_name
            print(f"successfully added {title} by {li[title]}")
            return 
    return 
        


def show(li):
    # همه اطلاعات موجود در دیکشنری    
    if not len(li):
        print("No Books in this library. Try adding more.")
        return
    if len(li):
        print("List of Books:\n")
        for key in li:
            print(key, li[key])


def search(li):
    #گشتن در دیکشنری 
    title = input("Enter book title: ").lower().capitalize()
    for key in li:
        if key == title:
            print(f"Author name: {li[key]}")
            return
        else:
            continue
    print("Not found.")
    

lib = {}
#اگر دیکشنری گلوبال هم بود چیز خوبی میشد
print("WELCOME TO THIS LIBRARY".center(60, '*'))
while True:
    usr_inpt = input("Please pick an option:\n1.Add a new book\n2.Show book info\n3.Search for a book\n4.exit")

#انتخابای کاربر رو مپ میکنیم
    if usr_inpt == '4':
        break
    if usr_inpt == '1':
       add(lib)
    if usr_inpt == '2':
        show(lib)
    if usr_inpt == '3':
        search(lib)
    else:
        print("invalid input. try again.")
```
