```python
### GUESS THE NUMBER GAME ###
import random
lost = True
pc_pick = random.randint(1, 20)
usr_pick = input("Guess what number I picked? ")
for i in range(5):
    if usr_pick.isdigit():
        usr_pick = int(usr_pick)
        if pc_pick == usr_pick:
            lost = False
            break
        elif pc_pick < usr_pick:
            usr_pick = input("Guess lower".center(40, "-"))
            continue
        else:
            usr_pick = input("Guess higher".center(40, "-"))
            continue
    else:
        usr_pick = input("invalid input. try again.")
        i -= 1
if lost:
    print("Oh no! No more guesses left!" + f"\nThe number was: {pc_pick}")
else:
    usr_pick = input("You got it!".center(40, "-"))
```

```python
###PERSONAL LIBRARY MANAGEMENT SYSTEM###


def add(li):
    # افزودن عنوان و نام نویسنده
    print("You are in 'add' interface.")
    for i in range(3):
        #سه بار می تونید تلاش کنید
        title = input("Enter book title: ").lower().capitalize()
        if not title.replace(" ", "").isalnum():
            print("English letters only.")
        elif title in li:
            print("Book already exists.")
            return 
        else: 
            break

    for i in range(3):
        auth_name = input("Enter name of the author: ").lower().capitalize()
        if not auth_name.replace(" ", "").isalpha():
            print("English letters only.")
        else:
            li[title] = auth_name
            print(f"successfully added {title} by {li[title]}")
            return 
    return 
        

def show(li):
    # همه اطلاعات موجود در دیکشنری    
    if not len(li):
        print("No Books in this library. Try adding more.")
        return
    if len(li):
        print("List of Books:\n")
        for key in li:
            print(key, li[key])


def search(li):
    #گشتن در دیکشنری 
    title = input("Enter book title: ").lower().capitalize()
    for key in li:
        if key == title:
            print(f"Author name: {li[key]}")
            return
        else:
            continue
    print("Not found.")
    

lib = {"The hunch back of notre dame":"Vitor Hugo", "Les Miserables":"Victor Hugo", "Harry potter":"J.K. Rowling"}

#اگر دیکشنری گلوبال هم بود چیز خوبی میشد
print("WELCOME TO THIS LIBRARY".center(60, '*'))
while True:
    usr_inpt = input("Please pick an option:\n1.Add a new book\n2.Show book info\n3.Search for a book\n0.exit")

#انتخابای کاربر رو مپ میکنیم
    if usr_inpt == '0':
        break
    if usr_inpt == '1':
       add(lib)
    if usr_inpt == '2':
        show(lib)
    if usr_inpt == '3':
        search(lib)
    elif not usr_inpt.isalnum():
        print("invalid input. try again.")
```

```python
#==========توابع کمکی============
def get_p_name(): #دریافت نام کالا تابعی هست که زیاد استفاده میشه در طول برنامه
    for _ in range(3): #سه بار فرصت
        inpt_pname = input("Enter product name: ").lower()

        if inpt_pname.replace(" ", "").isalnum():
            return inpt_pname #اعتبارسنجی استرینگ - ممانعت از کرکترای خاص
        print("No special characters.")
    print("invalid input.\n")


def get_p_quantity(p_name):
    for _ in range(3):
        quant = input(f"Enter quantity for {p_name}: ").strip()

        if quant.isdigit(): #دریافت تعداد
            return quant
        print("Enter positive english numbers only")
    print("invalid input\n")
    return 0


def add_to_existing(p_name, li): #پردازش اضافه کردن محصول'
    print(f"Product already registered.\n{p_name} : {li[p_name]} ")
    want = input("Want to add more?\n"
                "Press any key to add.\n"
                "Press 'enter' to cancel. ")
    if not want:
        return
    quant = get_p_quantity(p_name)
    if quant:
        li[p_name] += int(quant)
        print(f"successfully added {p_name} : {li[p_name]}")
        return


def add_new_p(p_name, li): #پردازش ایجاد محصول
    quant = get_p_quantity(p_name)
    if not quant:
        return
    else:
        li[p_name] = int(quant)
        print(f"successfully added {p_name} : {li[p_name]}")
        return

#==========توابع اصلی ===========
def add(li): #  افزودن کالا
    print("\n" + "'add' interface".center(50, '-') + "\n")

    p_name = get_p_name()
    if not p_name:
        print("\nSorry you have to start over\n")
        return
    
    if p_name in li: #برای محصول کلید موجوده
        add_to_existing(p_name, li)
        return
    elif p_name not in li:
        add_new_p(p_name, li)
        return
    else:
        print("Too many invalid attempts. Aborting...")


def sell(li):
    f_message = "\nPress any key + 'enter' to try again\nPress 'enter' to exit. "
    print("\n" + "'sell' interface".center(40, '-') + "\n")
    p_name = search(li, flag = False)
    
    for _ in range(3):
        quant = get_p_quantity(p_name)
        if quant is None:
            break
        if int(quant) > li[p_name]:
            cont = input(f"\nInsufficient stock. Only {li[p_name]} available." + f_message)
            if not cont:
                return li
            continue
        li[p_name] -= int(quant)
        print(f"Sold {quant} {p_name}(s). Remaining: {li[p_name]}")
        if li[p_name] == 0:
            del li[p_name]
        return li

    print("\nToo many invalid attempts. Aborting...")
    return
            

def search(li, flag):
    if flag:
        print("\n" + "'search' interface".center(40, '-') + "\n")
    f_message = "\nPress any key + 'enter' to try again\nPress 'enter' to exit. "
    for _ in range(3):
        p_name = get_p_name()
        if not p_name:
            print("\nSorry you have to start over\n")
            return
        elif p_name not in li:
            cont = input("Product not found." + f_message)
            if cont: 
                continue
            else: 
                return
        else:
            break
    
    print(f"{li[p_name]} {p_name}(s) in stock.\n")
    return p_name


def show(li):
    for i in li.items():
        print(str(i).center(50))


def save(li): #به روز رساتی فایل قبلی به جای جایگزین کردن اگر نامگذاری یکسان باشه البته
    saved = {}
    try:
        with open(usr_inpt, "r") as f:
            for line in f:
                if not line.strip():
                    continue
                p_name, q = line.strip().split(" - ")
                saved[p_name] = int(q)
    except FileNotFoundError:
        pass

    saved.update(li)

    with open("inventory.txt", "w") as f:
        for name, q in saved.items():
            f.write(f"{name} - {q}\n")
    print("\nsuccessfully saved your current inventory.")


def report(li):

    if not li:
        print("Inventory is empty.")
        return

    print(f"\nNumber of products: {len(li)}")

    total = sum(li.values())
    print(f"Total quantity: {total}")

    max_q = max(li.values())
    max_items = [name for name, q in li.items() if q == max_q] #شاید چند آیتم مساوی باشند
    print(f"Max stock: {', '.join(max_items)} ({max_q})")

    min_q = min(li.values())
    min_items = [name for name, q in li.items() if q == min_q]
    print(f"Min stock: {', '.join(min_items)} ({min_q})")

    return

inv = {}
lazy = input("r u lazy? (use sample inv) press 'enter' to skip: ")
if lazy:
    inv = {"lotion" : 400, "shampoo" : 100, "malafe" : 200, "soap" : 400, "mattress" : 50, "pillow" : 50, "slippers" : 30, "2 in 1" : 23}
while True: #UI اولیه
    usr_inpt = input("\nPlease pick an option:"
                    "\n1.Register a new product\n"
                    "2.Sell a product\n"
                    "3.Search for a product\n"
                    "4.Show products\n"
                    "5.Save\n"
                    "6.Get a report on products"
                    "\npress 0 to exit. ").strip()
    if usr_inpt == '0':
        break
    elif usr_inpt == '1':
        add(inv)
    elif usr_inpt == '2':
        sell(inv)
    elif usr_inpt == '3':
        search(inv, flag = True)
    elif usr_inpt == '4':
        show(inv)
    elif usr_inpt == '5':
        save(inv)
    elif usr_inpt == '6':
        report(inv)
    else:
        print("Invalid input. Try again. ")
```
